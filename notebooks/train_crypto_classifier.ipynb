# OpenCryptoDetect — Machine Learning Training Workflow

This notebook provides a reproducible workflow for generating dataset features, training the function-level cryptographic classifier, evaluating multi-architecture generalization metrics, and exporting the production model package.

All core logic resides inside the `opencryptodetect` package modules rather than ad-hoc notebook cells.

## 1. Environment Setup

## 2. Generate Feature Dataset across Cross-Architecture Targets

```python
from pathlib import Path
from dataset.generator.generate_corpus import generate_training_dataset

dataset_dir = Path("dataset/metadata")
dataset_file = generate_training_dataset(output_dir=dataset_dir, num_samples=200)
print(f"Dataset generated successfully: {dataset_file}")
```

## 3. Train Classifier (Random Forest / Gradient Boosting)

```python
from opencryptodetect.ml.classifier import train_classifier

model_path = Path("models/model_v1.joblib")
manifest = train_classifier(
    dataset_path=dataset_file,
    model_output_path=model_path,
    algorithm="rf"
)
print(f"Trained model version: {manifest.model_version}")
print(f"Validation accuracy: {manifest.accuracy:.4f}")
print(f"Weighted F1: {manifest.f1_weighted:.4f}")
```

## 4. Comprehensive Evaluation across Classes

```python
from opencryptodetect.ml.classifier import evaluate_classifier

evaluate_classifier(model_path=model_path, test_dataset_path=dataset_file)
```

## 5. Export Model with Manifest and Checksum

```python
from opencryptodetect.ml.model_registry import export_model_package

export_dir = Path("models")
manifest_file = export_model_package(model_path=model_path, output_dir=export_dir, manifest=manifest)
print(f"Exported model package with manifest: {manifest_file}")
```
